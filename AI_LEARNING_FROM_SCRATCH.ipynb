````python
size=int(input("Size of the Room:"))
bedrooms=int(input("No of Bedrooms:"))
age=int(input("Age of the House:"))

w1=int(input("Weight of Size:"))
w2=int(input("Weight of Bedrooms:"))
w3=int(input("Weight of Age:"))
b=int(input("Bias:"))

output=(size*w1)+(bedrooms*w2)+(age*w3)+b
print("Predicted House Price:",output)
````

Step1: Buid Simple AI that can learn

````python
#training Data
x=10
actual=50

#initial Values (random Guess)
weight=2
bias=0

#learning Rate
lr=0.1

for epoch in range(10):
  #prediction
  prediction=(x*weight)+bias

  print("Prediction:",prediction)
  print("Actual:",actual)

  #error
  error=actual-prediction

  print("Error   :",error)

  #update the weight
  weight= weight+(lr*error)
  print("New Weight:",weight)
````

Neural Network


````python
#inputs
x1=10
x2=6

#weights for neuron 1
w11=0.5
w12=0.8
b1=0

#weigth for neron 2
w21=0.2
w22=0.4
b2=0

#hidden layer
h1=(x1*w11)+(x2*w12)+b1
h2=(x1*w21)+(x2*w22)+b2

print("Hidden Layer1:",h1)
print("Hidden Layer2:",h2)


#output neuron weights
wo1=-6
wo2=0.9
bo=0

output=(h1*wo1)+(h2*wo2)+bo
print("Final output:",output)

#ReLU calculation
if output>0:
  output=output

else:
  output=0

print("After ReLU:",output)
````

````python
#Real backpropagation
#One training example
x=10
actual=50

#Initial random weights
w1=2
w2=3

#Learning rate
lr=0.001          # based on this the leraning rate take small or big
#-------------Forwared Pass ------------------------
hidden=x*w1
prediction=hidden*w2
loss=(actual-prediction)**2
print("Hidden:",hidden)
print("Prediction:",prediction)
print("Loss:",loss)

# #-----------Backward Pass----------------------
dLoss_dprediction=2*(prediction-actual)
dpredcition_dw2=hidden
dprediction_dhidden=w2
dhidden_dw1=x

gradient_w2=dLoss_dprediction*dpredcition_dw2
gradient_w1=dLoss_dprediction*dprediction_dhidden*dhidden_dw1

#-----------Update-------------------------

w1=w1-lr*gradient_w1
w2=w2-lr*gradient_w2

print("New weights:",w1,w2)
````

# **WHAT IS TENSORS?**



A **tensor** is a multi-dimensional array used to store and process data in deep learning. It is the basic data structure used in frameworks like **PyTorch** and **TensorFlow**.

### Types of Tensors

**0D Tensor (Scalar)** – A single value

```python
5
```

**1D Tensor (Vector)** – A list of values

```python
[10, 20, 30]
```

**2D Tensor (Matrix)** – Rows and columns

```python
[[1, 2, 3],
 [4, 5, 6]]
```

**3D Tensor** – A collection of matrices

```python
[
 [[1, 2],
  [3, 4]],

 [[5, 6],
  [7, 8]]
]
```

### Why are tensors used?

* Store data such as images, text, and audio.
* Store neural network weights and biases.
* Perform fast mathematical operations on CPUs and GPUs.

### Real-world Examples

* A grayscale image → **2D tensor** (Height × Width)
* A color image → **3D tensor** (Height × Width × Channels)
* A batch of images → **4D tensor** (Batch Size × Height × Width × Channels)

### Creating a Tensor in PyTorch

```python
import torch

tensor = torch.tensor([[1, 2],
                       [3, 4]])

print(tensor)
```

**Output**

```
tensor([[1, 2],
        [3, 4]])
```

### Key Point

A **tensor is simply a container of numbers with one or more dimensions**. Deep learning models use tensors to represent inputs, outputs, weights, and intermediate computations.

---

## 1. Dataset

A **Dataset** contains all the training samples (inputs and labels). It stores the complete data that will be used to train the neural network.

**Example**

```python
Images = [img1, img2, img3, ..., img1000]
Labels = [0, 1, 1, ..., 2]
```

The dataset itself does **not** feed data to the model. It only stores the data.

---

## 2. DataLoader

A **DataLoader** loads data from the dataset in small groups called **batches**. This makes training faster and uses less memory.

**Example**

```python
Dataset Size = 1000 images
Batch Size = 32

DataLoader →
Batch 1 → Images 1–32
Batch 2 → Images 33–64
Batch 3 → Images 65–96
...
```

---

## 3. Batch of Tensors

The DataLoader converts each batch into **tensors**, which are the input format required by PyTorch.

**Example**

```python
Batch Size = 32

Input Tensor Shape:
(32, 3, 224, 224)

32 → Number of images
3  → RGB channels
224 × 224 → Image size
```

Each batch is passed to the neural network.

---

## 4. Neural Network

The neural network processes the input tensors by passing them through multiple layers.

```
Input Tensor
      │
      ▼
Hidden Layer 1
      │
      ▼
Hidden Layer 2
      │
      ▼
Output Layer
```

The network learns patterns by adjusting its weights.

---

## 5. Prediction

The neural network produces predictions for every sample in the batch.

**Example**

```python
True Labels

[0, 2, 1, 0]

Predicted Labels

[0, 1, 1, 2]
```

These predictions are compared with the actual labels.

---

## 6. Loss Function

The **Loss Function** measures how far the predictions are from the correct labels.

* Small loss → Good prediction
* Large loss → Poor prediction

**Example**

```python
Prediction = Cat
Actual = Dog

Loss = High
```

```python
Prediction = Dog
Actual = Dog

Loss = Low
```

The goal of training is to **minimize the loss**.

---

## 7. Backpropagation

Backpropagation calculates how much each weight contributed to the loss.

It computes the gradients using the **chain rule** and sends the error backward through the network.

```
Loss
  ▲
  │
Output Layer
  ▲
  │
Hidden Layer
  ▲
  │
Input Layer
```

The gradient tells the network:

> "How should each weight change to reduce the loss?"

---

## 8. Optimizer

The optimizer updates the weights using the gradients computed during backpropagation.

Weight update formula:

[
\boxed{
\text{New Weight}=
\text{Old Weight}
-----------------

\text{Learning Rate}
\times
\text{Gradient}
}
]

After updating the weights, the network is ready to process the next batch.

---

### Key Idea

## **What is Underfitting?**



**Underfitting** occurs when a machine learning model is **too simple to learn the underlying patterns in the training data**. As a result, it performs poorly on both the training data and new (test) data.

### Characteristics

* The model fails to capture important patterns in the data.
* Training accuracy is low.
* Testing accuracy is also low.
* Training loss and validation loss remain high.

### Example

Suppose you want to predict house prices using only the number of bedrooms.

```text
Features Used:
- Number of bedrooms

Ignored Features:
- Location
- Area
- Age of the house
- Nearby facilities
```

Since the model uses very little information, it cannot learn the true relationship between the features and the price. This is an example of **underfitting**.

### Graph

```text
Actual Data

        •
     •      •
   •           •
 •               •

Model Prediction (Too Simple)

-------------------------------
```

The straight line does not capture the pattern in the data.

### Causes of Underfitting

* Model is too simple.
* Too few training epochs.
* Important features are missing.
* Excessive regularization.
* Insufficient model capacity.

### How to Reduce Underfitting

* Use a more complex model.
* Train for more epochs.
* Add more relevant features.
* Reduce excessive regularization.
* Increase the number of neurons or layers in a neural network.

### Summary

* **Underfitting** = Model is **too simple**.
* **Training Accuracy:** Low
* **Testing Accuracy:** Low
* **Training Loss:** High
* **Validation Loss:** High

**Easy way to remember:**

> **Underfitting = The model has not learned enough from the training data.**


# **What is Overfitting?**



**Overfitting** occurs when a machine learning model **learns the training data too well, including its noise and unnecessary details**. As a result, it performs very well on the training data but poorly on new (test) data.

### Characteristics

* The model memorizes the training data instead of learning general patterns.
* Training accuracy is very high.
* Testing accuracy is low.
* Training loss is very low, while validation loss is high.

### Example

Suppose you want to predict house prices.

```text
Training Data Features:
- Number of bedrooms
- Area
- Location
- Age of the house

The model also memorizes:
- Random noise
- Outliers
- Specific training examples
```

The model gives excellent predictions on the training data but fails to predict prices accurately for new houses.

### Graph

```text
Actual Data

        •
     •      •
   •           •
 •               •

Overfitted Model

      /\/\__/\/\___/\_
```

The curve passes through almost every training point, including noise, instead of capturing the general trend.

### Causes of Overfitting

* Model is too complex.
* Too many training epochs.
* Small training dataset.
* Too many features.
* Lack of regularization.

### How to Reduce Overfitting

* Collect more training data.
* Use a simpler model.
* Apply regularization (L1/L2).
* Use Dropout in neural networks.
* Use Early Stopping.
* Perform data augmentation.
* Train for an appropriate number of epochs.

### Summary

* **Overfitting** = Model **memorizes** the training data instead of learning general patterns.
* **Training Accuracy:** High
* **Testing Accuracy:** Low
* **Training Loss:** Low
* **Validation Loss:** High

**Easy way to remember:**

> **Overfitting = The model learns the training data too specifically and cannot generalize well to new data.**


# **PyTorch Optimizers**



An **optimizer** is an algorithm that **updates the weights and biases of a neural network to minimize the loss function**. After backpropagation computes the gradients, the optimizer uses those gradients to adjust the model parameters.

### Training Flow

```text
Input
   │
   ▼
Neural Network
   │
   ▼
Prediction
   │
   ▼
Loss Function
   │
   ▼
Backpropagation (Computes Gradients)
   │
   ▼
Optimizer (Updates Weights)
   │
   ▼
Repeat
```

* **Learning Rate:** Controls the size of the weight update.
* **Gradient:** Indicates the direction and magnitude to change the weight.

## Common PyTorch Optimizers

### 1. SGD (Stochastic Gradient Descent)

Updates the weights using the gradient computed from the current batch.

**Formula**
image.png

where:

* (w) = Weight
* (\eta) = Learning rate
* (L) = Loss

**PyTorch Example**

```python
import torch.optim as optim

optimizer = optim.SGD(model.parameters(), lr=0.01)
```

**Advantages**

* Simple and easy to understand.
* Uses less memory.
* Works well for many problems.

**Disadvantages**

* Can converge slowly.
* May get stuck in local minima.

---

### 2. SGD with Momentum

Momentum helps the optimizer move faster in the correct direction by considering previous updates.

**PyTorch Example**

```python
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)
```

**Advantages**

* Faster convergence than standard SGD.
* Reduces oscillations during training.

---

### 3. Adam (Adaptive Moment Estimation)

Adam adapts the learning rate for each parameter using estimates of the first and second moments of the gradients.

**PyTorch Example**

```python
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

**Advantages**

* Fast convergence.
* Works well for most deep learning tasks.
* Default choice for many applications.

**Disadvantages**

* Uses more memory than SGD.
* May not always generalize as well as SGD.

---

### 4. AdamW

AdamW is an improved version of Adam that applies weight decay correctly.

**PyTorch Example**

```python
optimizer = optim.AdamW(model.parameters(), lr=0.001)
```

**Advantages**

* Better regularization.
* Commonly used for Transformer and NLP models.

---

### 5. RMSprop

RMSprop adjusts the learning rate based on recent gradients.

**PyTorch Example**

```python
optimizer = optim.RMSprop(model.parameters(), lr=0.001)
```

**Advantages**

* Works well for recurrent neural networks (RNNs).
* Handles non-stationary objectives effectively.

---

## Comparison

| Optimizer      | Speed    | Memory | Best Use                        |
| -------------- | -------- | ------ | ------------------------------- |
| SGD            | Moderate | Low    | General deep learning           |
| SGD + Momentum | Fast     | Low    | Faster convergence than SGD     |
| Adam           | Fast     | Medium | Most deep learning applications |
| AdamW          | Fast     | Medium | Transformers, NLP               |
| RMSprop        | Fast     | Medium | RNNs and sequence models        |

---

## Typical Training Loop

```python
import torch.optim as optim

optimizer = optim.Adam(model.parameters(), lr=0.001)

for inputs, labels in dataloader:

    optimizer.zero_grad()      # Clear old gradients

    outputs = model(inputs)    # Forward pass

    loss = criterion(outputs, labels)

    loss.backward()            # Compute gradients

    optimizer.step()           # Update weights
```

---

## Key Points

* An **optimizer updates the model's weights** to reduce the loss.
* **Backpropagation computes the gradients**; the optimizer uses those gradients.
* **SGD** is simple and memory-efficient.
* **Adam** is the most commonly used optimizer because it converges quickly and performs well across many tasks.
* **AdamW** is preferred for many modern Transformer-based models.

# **Interview question Answers**

**While pruning, how do we decide which weights are important and which are less important?**

---



One approach is to analyze how much each weight affects the model's loss. Backpropagation computes gradients for all weights, showing how sensitive the loss is to each weight. Weights with consistently small gradients have little influence on the loss and are considered less important. These weights can often be pruned with minimal impact on the model's performance.
